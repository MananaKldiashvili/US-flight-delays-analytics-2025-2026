### US Flight Delays & Airline Reliability Analytics (Period:2025-01 through 2026-05)

ყველა სვეტის ამოღება PascalCase ალიასებით და მოკლე განმარტებებით

```sql
-- 1. ნედლი ფაქტების ცხრილის (airline_airport_monthly) სვეტების ამოღება

SELECT 
    Mnth.year AS Year, -- ფრენის განხორციელების წელი (2025 ან 2026).
    Mnth.month AS Month, -- ფრენის განხორციელების თვე (1-დან 12-მდე).
    Mnth.period AS Period, -- წლისა და თვის კომბინირებული ტექსტური ფორმატი (YYYY-MM).
    Mnth.carrier AS CarrierCode, -- ავიაკომპანიის 2-სიმბოლოიანი IATA კოდი (მაგ. AA, DL).
    Mnth.carrier_name AS CarrierName, -- ავიაკომპანიის სრული ოფიციალური დასახელება.
    Mnth.airport AS AirportCode, -- აეროპორტის 3-ასოიანი IATA კოდი (მაგ. ATL, JFK).
    Mnth.airport_name AS AirportName, -- აეროპორტისა და ქალაქის სრული დასახელება.
    Mnth.arr_flights AS TotalRepresentedFlights, -- იმ თვეში განხორციელებული ჯამური ფრენების რაოდენობა.
    Mnth.arr_del15 AS DelayedFlights15Plus, -- 15 წუთზე მეტი ხანგრძლივობით დაგვიანებული ფრენების რაოდენობა.
    Mnth.arr_cancelled AS CancelledFlights, -- გაუქმებული ფრენების ჯამური რაოდენობა.
    Mnth.arr_diverted AS DivertedFlights, -- სხვა აეროპორტში გადამისამართებული ფრენების რაოდენობა.
    Mnth.arr_delay AS TotalDelayMinutes, -- დაგვიანების ჯამური ხანგრძლივობა წუთებში.
    Mnth.carrier_ct AS CarrierDelayCount, -- ავიაკომპანიის მიზეზით დაგვიანებული ფრენების რაოდენობა.
    Mnth.weather_ct AS WeatherDelayCount, -- ამინდის მიზეზით დაგვიანებული ფრენების რაოდენობა.
    Mnth.nas_ct AS NasDelayCount, -- აეროპორტის/NAS-ის მიზეზით დაგვიანებული ფრენების რაოდენობა.
    Mnth.security_ct AS SecurityDelayCount, -- უსაფრთხოების მიზეზით დაგვიანებული ფრენების რაოდენობა.
    Mnth.late_aircraft_ct AS LateAircraftDelayCount, -- წინა რეისის დაგვიანებით გამოწვეული დაგვიანებული ფრენების რაოდენობა.
    Mnth.carrier_delay AS CarrierDelayMinutes, -- ავიაკომპანიის მიზეზით გამოწვეული დაგვიანების წუთები.
    Mnth.weather_delay AS WeatherDelayMinutes, -- ამინდის მიზეზით გამოწვეული დაგვიანების წუთები.
    Mnth.nas_delay AS NasDelayMinutes, -- NAS-ის მიზეზით გამოწვეული დაგვიანების წუთები.
    Mnth.security_delay AS SecurityDelayMinutes, -- უსაფრთხოების მიზეზით გამოწვეული დაგვიანების წუთები.
    Mnth.late_aircraft_delay AS LateAircraftDelayMinutes, -- წინა რეისის დაგვიანებით გამოწვეული დაგვიანების წუთები.
    Mnth.delay_rate_pct AS DelayRatePct, -- დაგვიანებული ფრენების პროცენტული წილი (% = (arr_del15 / arr_flights) * 100).
    Mnth.cancellation_rate_pct AS CancellationRatePct, -- გაუქმებული ფრენების პროცენტული წილი (% = (arr_cancelled / arr_flights) * 100).
    Mnth.avg_delay_minutes_per_delayed_flight AS AvgDelayMinutesPerDelayed, -- საშუალო დაგვიანება წუთებში ერთ დაგვიანებულ ფრენაზე (= arr_delay / arr_del15).
    Mnth.carrier_delay_share_pct AS CarrierDelaySharePct, -- ავიაკომპანიის ბრალეულობის წილი დაგვიანებულ წუთებში (% = (carrier_delay / arr_delay) * 100).
    Mnth.weather_delay_share_pct AS WeatherDelaySharePct, -- ამინდის ბრალეულობის წილი დაგვიანებულ წუთებში (% = (weather_delay / arr_delay) * 100).
    Mnth.nas_delay_share_pct AS NasDelaySharePct, -- NAS-ის ბრალეულობის წილი დაგვიანებულ წუთებში (% = (nas_delay / arr_delay) * 100).
    Mnth.security_delay_share_pct AS SecurityDelaySharePct, -- უსაფრთხოების ბრალეულობის წილი დაგვიანებულ წუთებში (% = (security_delay / arr_delay) * 100).
    Mnth.late_aircraft_delay_share_pct AS LateAircraftDelaySharePct -- წინა რეისის დაგვიანების წილი დაგვიანებულ წუთებში (% = (late_aircraft_delay / arr_delay) * 100).
FROM getdata.raw.airline_airport_monthly AS Mnth;
```

```sql
-- 2. ავიაკომპანიების ოფიციალური ცნობარის (airline_rankings) სვეტების ამოღება

SELECT 
    AlR.carrier AS CarrierCode, -- ავიაკომპანიის 2-სიმბოლოიანი IATA კოდი.
    AlR.carrier_name AS CarrierName, -- ავიაკომპანიის ოფიციალური დასახელება.
    AlR.delay_rate_rank AS AirlineOfficialRank -- ავიაკომპანიის ოფიციალური ადგილი რეიტინგში აშშ-ში.
FROM getdata.raw.airline_rankings_2025_2026 AS AlR;
```

```sql
-- 3. აეროპორტების ოფიციალური ცნობარის (airport_rankings) სვეტების ამოღება

SELECT 
    ApR.airport AS AirportCode, -- აეროპორტის 3-ასოიანი IATA კოდი.
    ApR.airport_name AS AirportName, -- აეროპორტისა და ქალაქის სრული დასახელება.
    ApR.delay_rate_rank AS AirportOfficialRank -- აეროპორტის ოფიციალური რანგი (ადგილი რეიტინგში) აშშ-ში.
FROM getdata.raw.airport_rankings_2025_2026 AS ApR;
```
